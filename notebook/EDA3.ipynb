```python
import pandas as pd

df = pd.read_csv('../data/processed/shopee_gold_dataset_absa_5_aspects.csv')

print("Kích thước:", df.shape)
print("\nCác cột:")
print(df.columns.tolist())

print("\n5 dòng đầu:")
display(df.head())
```

Thống kê số lượng + tỷ lệ của 5 aspect

```python
aspects = [
    'label_quality',
    'label_price',
    'label_delivery',
    'label_packaging',
    'label_service'
]

for col in aspects:
    print("\n" + "="*50)
    print(col.upper())
    print("="*50)

    counts = df[col].value_counts()
    percentages = df[col].value_counts(normalize=True) * 100

    print(pd.DataFrame({
        'Số lượng': counts,
        'Tỷ lệ (%)': percentages.round(2)
    }))
```

```python
for col in aspects:
    print(f"{col}: {df[col].ne('none').sum()} review có đề cập aspect")
```

```python
df['aspect_count'] = (df[aspects] != 'none').sum(axis=1)

print(df['aspect_count'].value_counts().sort_index())
```

```python
display(
    df[df['aspect_count'] >= 3][
        ['review_text'] + aspects
    ].head(10)
)
```
