```python
import pandas as pd
import os

# Tạo thư mục sample ở ngoài root
os.makedirs('../data/sample', exist_ok=True)

# Đọc file dataset (lùi 1 cấp ra ngoài, vào data/processed)
df = pd.read_csv('../data/processed/shopee_gold_dataset_absa_5_aspects.csv')

# Lấy ngẫu nhiên 50 dòng và xuất ra file mới
df.sample(50, random_state=42).to_csv('../data/sample/sample_reviews.csv', index=False)

print("Tạo sample data thành công!")
```
